# CENÁRIOS

```python
cenarios = {
    "Operação Normal": {
        "sensor_termico": 90,
        "sensor_optico": 95,
        "buffer_imagens": 30,
        "precisao_geolocalizacao": 1.2,
        "energia_disponivel": 80
    },

    "Energia Crítica": {
        "sensor_termico": 90,
        "sensor_optico": 95,
        "buffer_imagens": 30,
        "precisao_geolocalizacao": 1.2,
        "energia_disponivel": 15
    },

    "Temperatura Crítica": {
        "sensor_termico": 50,
        "sensor_optico": 95,
        "buffer_imagens": 30,
        "precisao_geolocalizacao": 1.2,
        "energia_disponivel": 80
    },

    "Buffer Elevado": {
        "sensor_termico": 90,
        "sensor_optico": 95,
        "buffer_imagens": 95,
        "precisao_geolocalizacao": 1.2,
        "energia_disponivel": 80
    },

    "Falha de Geolocalização": {
        "sensor_termico": 90,
        "sensor_optico": 95,
        "buffer_imagens": 30,
        "precisao_geolocalizacao": 8,
        "energia_disponivel": 80
    }
}
```


# MOTOR DE ALERTAS

```python
def avaliar(dados):

    alertas = []

    if dados["energia_disponivel"] < 20:
        alertas.append("🔴 CRÍTICO: Energia abaixo de 20%")

    if dados["sensor_termico"] < 60:
        alertas.append("🔴 CRÍTICO: Sensor térmico degradado")

    if dados["buffer_imagens"] > 85:
        alertas.append("🟠 ALERTA: Buffer acima de 85%")

    if dados["precisao_geolocalizacao"] > 5:
        alertas.append("🟠 ALERTA: Precisão geográfica comprometida")

    if not alertas:
        alertas.append("🟢 Operação normal")

    return alertas
```

# EXECUÇÃO

```python
for nome, dados in cenarios.items():

    print("\n")
    print("=" * 60)
    print(nome)
    print("=" * 60)

    for alerta in avaliar(dados):
        print(alerta)
```

# Conclusão

O sistema conseguiu identificar corretamente todas as anomalias simuladas.

A solução demonstra como técnicas simples de monitoramento podem ser combinadas com IA generativa para apoiar operações espaciais voltadas ao monitoramento ambiental.
